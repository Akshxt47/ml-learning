```python
import pandas as pd 
import matplotlib.pyplot as plt

copd = pd.read_csv('COPD.txt', sep='\t', header=None)
control = pd.read_csv('Control.txt', sep='\t', header=None)

print(copd.shape)
print(control.shape)
```

```python

plt.plot(copd[3], label='COPD patient (D01)')
plt.plot(control[3], label='Healthy person (C01)')

plt.xlabel('Time (reading number)')
plt.ylabel('Sensor 4 signal')
plt.title('Sensor 4 response: COPD vs healthy')
plt.legend()
plt.show()
```

```python
for i in range(20):
    plt.plot(copd[3 + 16*i], color='red', alpha=0.4)

for i in range(10):
    plt.plot(control[3 + 16*i], color='blue', alpha=0.4)

plt.xlabel('Time (reading number)')
plt.ylabel('Sensor 4 signal')
plt.title('Sensor 4, test 1: COPD (red) vs healthy (blue)')
plt.show()
```
